```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-classification", model="disham993/electrical-classification-ModernBERT-base")
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained("disham993/electrical-classification-ModernBERT-base")
model = AutoModelForSequenceClassification.from_pretrained("disham993/electrical-classification-ModernBERT-base")
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline

model_name = "disham993/electrical-classification-ModernBERT-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
nlp = pipeline("text-classification", model=model, tokenizer=tokenizer)

text = "The new washing machine is efficient but produces a bit of noise."
classification_results = nlp(text)
print(classification_results)
```
